```python
import pandas as pd
from pathlib import Path

# Directory where this script lives
BASE_DIR = Path.cwd()
```

```python
file = pd.read_csv(BASE_DIR / "../Dataset/Labled_Data/Avi_5000_annotated.csv")
```

```python
file.head()
```

```python
labels = file['label_signals']
```

```python
dict_labels = {
    "attention_dysregulation":0,
    "anxious_affect":0,
    "autistic_trait_discussion":0,
    "emotional_instability":0,
    "depressive_affect":0,
    "trauma_stress":0,
    "crisis_self_harm":0,
    "no_clear_signal":0
}
```

```python
for label in labels:
    for sub_labels in label.split(","):
        dict_labels[sub_labels] +=1
    
print(dict_labels)
```
